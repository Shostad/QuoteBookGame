```python
textFile = open("Quote2.txt",'r',encoding='utf-8')
Authors = open("Authors.txt",'r',encoding='utf-8')
output = open("Server/POSTGRESScripts/Quote book data entry.sql",'w',encoding='utf-8')
```

```python
textRaw = ""
for x in textFile:
    textRaw += x
print(textRaw)

textSplit = textRaw.split("\n\n")
# print(textSplit)
```

```python
text = []
people = []
date = []
currentQuote = []

def concatStr(arr):
    out = ''
    for x in arr:
        out+=x
        out+=' '
    return out

for x in textSplit:
    currentQuote = x.split('"')
    print(currentQuote)
    text.append(currentQuote[1].replace('\'','\\\''))
    namesAndDate = currentQuote[2].split(' ')
    date.append(namesAndDate[-1])
    people.append(namesAndDate[0:-1])
    people[-1] = concatStr(people[-1]).lower()

# getting rid of empty strings
# for x in people:
#     for y in range(len(x)-1):
#         if x[y] == '':
#             x.remove(x[y])
# # print(currentQuote)
# print(date)
```

```python
print(date)
out = []
for x in date:
    temp = x.split('/')
    print(temp)
    if len(temp) < 3:
        raise ValueError("Incorrectly formatted date: '"+str(x)+"'")
    out.append('20'+temp[2]+'/'+temp[1]+'/'+temp[0])
print(out)
date = out
```

```python
AuthorList = []
temp = []
for x in Authors:
    temp = (x[0:-1].split(','))
    for y in temp:
        AuthorList.append(y)
    print(AuthorList)
```

```python
formattedData = []
for x in range(len(text)):
    # tempPeople = ''
    tempPeople = []
    for y in AuthorList:
        if y.lower() in people[x]:
            # tempPeople+=y
            # tempPeople+=','
            tempPeople.append(y)
    # tempPeople= tempPeople[0:-1]
    if tempPeople == []:
        raise ValueError("quote completed with empty authors"+str(text[x]))
    formattedData.append((text[x],tempPeople,date[x]))
print(formattedData)
```

```python
peopleStatement = ''
for x in AuthorList:
    baseText = '''insert into person (name,created_by) values ('''
    peopleStatement = peopleStatement + baseText+'\'' + x + "',1);\n"
print(peopleStatement)
```

```python
quoteStatement = ''
for x in formattedData:
    baseText = '''insert into quote (text,created_by) values ('''
    quoteStatement = quoteStatement + baseText+'$$' + x[0] + "$$,1);\n"
print(quoteStatement)
```
